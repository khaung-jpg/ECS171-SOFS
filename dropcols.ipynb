```python
import os
import pandas as pd
from harmonize import harmonize

os.makedirs("data/harmonized", exist_ok=True)

for year in [2017, 2018, 2019, 2020, 2021]:
    clean = pd.read_csv(f"data/clean/{year}_basic.csv", low_memory=False)
    h = harmonize(clean, year)
    h.to_csv(f"data/harmonized/{year}_droppedcols.csv", index=False)
    print(year, h.shape, "\n")
```
